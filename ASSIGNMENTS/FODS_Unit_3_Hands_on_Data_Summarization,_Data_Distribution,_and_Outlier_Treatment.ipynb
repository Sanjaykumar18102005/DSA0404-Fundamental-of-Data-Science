```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew

# 1. Generate a sample dataset of house prices
# Let's create a synthetic dataset that is slightly skewed to the right
np.random.seed(42) # for reproducibility
house_prices = np.concatenate([
    np.random.normal(loc=250000, scale=50000, size=700), # Majority of houses
    np.random.normal(loc=600000, scale=100000, size=200), # More expensive houses
    np.random.normal(loc=150000, scale=30000, size=100) # Less expensive houses
])

# Ensure all prices are positive
house_prices = house_prices[house_prices > 0]

# Create a Pandas Series for easy handling
house_prices_series = pd.Series(house_prices, name='House Prices')

print(f"Generated {len(house_prices_series)} sample house prices.")
print(f"Minimum Price: ${house_prices_series.min():,.2f}")
print(f"Maximum Price: ${house_prices_series.max():,.2f}")
print(f"Mean Price: ${house_prices_series.mean():,.2f}\n")

# 2. Plot the histogram of house prices
plt.figure(figsize=(10, 6))
sns.histplot(house_prices_series, bins=50, kde=True, color='skyblue', edgecolor='black')
plt.title('Distribution of House Prices')
plt.xlabel('House Price ($)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# 3. Calculate skewness
skewness_value = skew(house_prices_series)
print(f"\nSkewness of House Prices: {skewness_value:.4f}")
```

```python
import pandas as pd
import numpy as np

# 1. Create a sample dataset of student marks
# Let's include some potential outliers for demonstration
marks = [
    65, 70, 72, 75, 78, 80, 81, 83, 85, 88,
    90, 92, 95, 98, 100, # Regular marks
    30, 45, # Low outliers
    105, 110 # High outliers (e.g., bonus points, data entry errors)
]

marks_series = pd.Series(marks, name='Student Marks')

print("### Student Marks Data ###\n")
print(marks_series)
print("\n")

# 2. Calculate the Interquartile Range (IQR)
Q1 = marks_series.quantile(0.25)
Q3 = marks_series.quantile(0.75)
IQR = Q3 - Q1

print(f"First Quartile (Q1): {Q1}")
print(f"Third Quartile (Q3): {Q3}")
print(f"Interquartile Range (IQR): {IQR}\n")

# 3. Define outlier bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Lower Bound for Outliers: {lower_bound}")
print(f"Upper Bound for Outliers: {upper_bound}\n")

# 4. Detect outliers
outliers = marks_series[(marks_series < lower_bound) | (marks_series > upper_bound)]

if not outliers.empty:
    print("### Detected Outliers ###\n")
    print(outliers)
else:
    print("No outliers detected in the student marks using the IQR method.")

# Optional: Visualize with a box plot to confirm
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.boxplot(y=marks_series, palette='coolwarm')
plt.title('Box Plot of Student Marks with Outliers')
plt.ylabel('Marks')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create a sample sales dataset with some outliers
np.random.seed(42) # for reproducibility
sales_data = np.concatenate([
    np.random.normal(loc=1000, scale=200, size=90), # Regular sales
    np.random.normal(loc=5000, scale=500, size=5),  # High outliers
    np.random.normal(loc=100, scale=30, size=5)   # Low outliers
])

sales_series = pd.Series(sales_data, name='Daily Sales ($)')

print("### Original Sales Data (with potential outliers) ###\n")
display(sales_series.describe())

# Visualize original data
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(sales_series, bins=30, kde=True, color='purple', edgecolor='black')
plt.title('Original Sales Distribution')
plt.xlabel('Sales Amount ($)')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
sns.boxplot(y=sales_series, color='purple') # Changed palette to color
plt.title('Original Sales Box Plot')
plt.ylabel('Sales Amount ($)')
plt.tight_layout()
plt.show()

# 2. Calculate the Interquartile Range (IQR) for sales data
Q1_sales = sales_series.quantile(0.25)
Q3_sales = sales_series.quantile(0.75)
IQR_sales = Q3_sales - Q1_sales

print(f"\nFirst Quartile (Q1): {Q1_sales:.2f}")
print(f"Third Quartile (Q3): {Q3_sales:.2f}")
print(f"Interquartile Range (IQR): {IQR_sales:.2f}\n")

# 3. Define outlier bounds
lower_bound_sales = Q1_sales - 1.5 * IQR_sales
upper_bound_sales = Q3_sales + 1.5 * IQR_sales

print(f"Lower Bound for Outliers: {lower_bound_sales:.2f}")
print(f"Upper Bound for Outliers: {upper_bound_sales:.2f}\n")

# 4. Identify outliers
outliers_sales = sales_series[
    (sales_series < lower_bound_sales) |
    (sales_series > upper_bound_sales)
]

print("### Detected Outliers in Sales Data ###\n")
if not outliers_sales.empty:
    display(outliers_sales)
else:
    print("No outliers detected in the sales data using the IQR method.")

# 5. Remove outliers from the dataset
cleaned_sales_series = sales_series[
    (sales_series >= lower_bound_sales) &
    (sales_series <= upper_bound_sales)
]

print("\n### Cleaned Sales Data (Outliers Removed) ###\n")
display(cleaned_sales_series.describe())

# Visualize cleaned data
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(cleaned_sales_series, bins=30, kde=True, color='green', edgecolor='black')
plt.title('Cleaned Sales Distribution')
plt.xlabel('Sales Amount ($)')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
sns.boxplot(y=cleaned_sales_series, color='green') # Changed palette to color
plt.title('Cleaned Sales Box Plot')
plt.ylabel('Sales Amount ($)')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew

print("### Starting Titanic Dataset EDA ###\n")

# 1. Download and Load Titanic Dataset
try:
    df = sns.load_dataset('titanic')
    print("Titanic dataset loaded successfully.\n")
except Exception as e:
    print(f"Error loading dataset: {e}. Please ensure you have an internet connection or the dataset is available locally.")
    # Fallback if seaborn fails to load directly (though it's usually reliable)
    # You might consider providing a URL for direct download if this is a common issue
    df = pd.DataFrame() # Create an empty dataframe to prevent further errors if loading fails

if not df.empty:
    # Create a copy to work with, preserving the original if needed
    titanic_df = df.copy()

    print("### Dataset Information ###\n")
    # Display dataset information
    print("First 5 rows of the dataset:")
    display(titanic_df.head())
    print(f"\nDataset Shape: {titanic_df.shape[0]} rows, {titanic_df.shape[1]} columns")
    print("\nDetailed info:")
    titanic_df.info()

    print("\n### Missing Values Analysis ###\n")
    # Finding missing values
    missing_values = titanic_df.isnull().sum()
    missing_percentage = (titanic_df.isnull().sum() / len(titanic_df)) * 100
    missing_info = pd.DataFrame({
        'Missing Count': missing_values,
        'Missing Percentage': missing_percentage
    })
    display(missing_info[missing_info['Missing Count'] > 0].sort_values(by='Missing Count', ascending=False))

    # Handling missing values
    # 'age': Impute with median as it's numerical and can be skewed
    titanic_df['age'].fillna(titanic_df['age'].median(), inplace=True)
    # 'embarked': Impute with mode as it's categorical
    titanic_df['embarked'].fillna(titanic_df['embarked'].mode()[0], inplace=True)
    # 'deck': Too many missing values (77%), drop the column
    titanic_df.drop('deck', axis=1, inplace=True)
    # 'fare': Impute with median for any potential missing values (though typically none in titanic dataset)
    titanic_df['fare'].fillna(titanic_df['fare'].median(), inplace=True)

    print("\nMissing values after imputation/dropping:\n")
    display(titanic_df.isnull().sum().to_frame(name='Remaining Missing Count'))

    print("\n### Descriptive Statistics ###\n")
    # Computing descriptive statistics
    display(titanic_df.describe(include='all'))

    print("\n### Data Visualization ###\n")
    # Identify numerical and categorical columns for plotting
    numerical_cols = titanic_df.select_dtypes(include=np.number).columns.tolist()
    categorical_cols = titanic_df.select_dtypes(exclude=np.number).columns.tolist()

    # Plotting Histograms for numerical columns
    print("Histograms for Numerical Features:\n")
    plt.figure(figsize=(15, 10))
    for i, col in enumerate(numerical_cols):
        plt.subplot(3, 3, i + 1) # Adjust subplot grid as needed
        sns.histplot(titanic_df[col], kde=True, bins=30)
        plt.title(f'Distribution of {col}')
    plt.tight_layout()
    plt.show()

    # Plotting Box Plots for numerical columns
    print("\nBox Plots for Numerical Features (Outlier Detection Visualization):\n")
    plt.figure(figsize=(15, 10))
    for i, col in enumerate(numerical_cols):
        plt.subplot(3, 3, i + 1) # Adjust subplot grid as needed
        sns.boxplot(y=titanic_df[col])
        plt.title(f'Box Plot of {col}')
    plt.tight_layout()
    plt.show()

    print("\n### Outlier Detection and Removal (using IQR method) ###\n")

    def detect_and_remove_outliers_iqr(df, column):
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
        df_cleaned = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

        print(f"\nColumn: {column}")
        print(f"  Q1: {Q1:.2f}, Q3: {Q3:.2f}, IQR: {IQR:.2f}")
        print(f"  Lower Bound: {lower_bound:.2f}, Upper Bound: {upper_bound:.2f}")
        print(f"  Number of outliers detected: {len(outliers)}")
        if not outliers.empty:
            print("  Sample outliers:\n")
            display(outliers.head())
        return df_cleaned, outliers

    # Apply outlier detection and removal to key numerical columns
    cleaned_titanic_df = titanic_df.copy()
    columns_to_check_outliers = ['age', 'fare', 'sibsp', 'parch'] # Exclude 'pclass' as it's more categorical

    all_outliers = pd.DataFrame()

    for col in columns_to_check_outliers:
        initial_rows = len(cleaned_titanic_df)
        cleaned_titanic_df, current_outliers = detect_and_remove_outliers_iqr(cleaned_titanic_df, col)
        all_outliers = pd.concat([all_outliers, current_outliers])
        print(f"  {initial_rows - len(cleaned_titanic_df)} rows removed for outliers in '{col}'. Remaining rows: {len(cleaned_titanic_df)}")

    print(f"\nTotal rows before outlier removal: {len(titanic_df)}")
    print(f"Total rows after outlier removal: {len(cleaned_titanic_df)}")
    print(f"Number of unique outliers identified across selected columns: {len(all_outliers.drop_duplicates())}\n")

    # Visualize distributions after outlier removal
    print("Distributions after Outlier Removal:\n")
    plt.figure(figsize=(15, 10))
    for i, col in enumerate(columns_to_check_outliers):
        if col in cleaned_titanic_df.columns: # Check if column still exists after previous outlier removal steps
            plt.subplot(2, 2, i + 1)
            sns.histplot(cleaned_titanic_df[col], kde=True, bins=30, color='green')
            plt.title(f'Cleaned Distribution of {col}')
    plt.tight_layout()
    plt.show()

    print("\n### Saving Cleaned Dataset ###\n")
    # Saving the cleaned dataset
    output_filename = 'titanic_cleaned.csv'
    cleaned_titanic_df.to_csv(output_filename, index=False)
    print(f"Cleaned Titanic dataset saved to '{output_filename}'")

    print("\n### EDA Complete ###")
else:
    print("Cannot proceed with EDA as the dataset failed to load.")
```
